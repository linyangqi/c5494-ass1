1.1 accomplish a Fully Visible Sigmoid Belief Network (e.g., logistic
regression) for image generation on the binarized MNIST data.

1.2 accomplish a neural based autoregressive model (e.g., one or multiple
layers) for image generation on the binarized MNIST data.
You may need to first create binarized images yourself (based on MNIST). The generated images
do not have to be strictly 28x28 in size, you can choose an appropriate dimension based on your
model design. The number of training images is also up to you to decide but ensure there are
enough samples to train your autoregressive model, and you need to demonstrate your model’s
generation capability (e.g., at least 10 images are recommended for submission).

1.3  implement and train a GAN (you can choose your own encoder and
decoder architecture) for handwritten digit generation on the original MNIST dataset (not the
binarized version). For the training data, you may choose to use either a subset of the training set
or the entire training set, specify your choice (and the size of the subset, if applicable) in your
report. After training, the generator must be able to generate MNIST-like digit images, and you
need to submit some generated images along with a brief report on training stability (e.g., whether
mode collapse occurred).
